# Задача 4. Интернет-магазин (merge и concat)

Магазин работает через два филиала, и каждый хранит свои заказы в отдельном файле. Из-за сбоя синхронизации часть заказов попала в **оба** файла — это **один и тот же заказ** (все значения совпадают).

| файл | столбцы |
|---|---|
| `data/orders_msk.csv`, `data/orders_spb.csv` | `order_id` (номер заказа), `customer_id`, `product_id`, `quantity` (штук). **Один заказ — один товар** |
| `data/products.csv` | `product_id`, `category`, `unit_price` (цена за штуку, руб.) |
| `data/customers.csv` | `customer_id`, `segment` (`Standard`, `Premium`, `VIP`), `city` |

Выручка заказа = `quantity × unit_price`. **Заказы с товаром, которого нет в `products.csv`, в расчётах выручки не участвуют**, но в подсчёте количества заказов — участвуют, если не сказано иное.

```python
# Загрузите 4 датасета


msk = pd.read_csv('data/orders_msk.csv')
spb = pd.read_csv('data/orders_spb.csv')

products = pd.read_csv('data/products.csv')
customers = pd.read_csv('data/customers.csv')
```

```python
msk.info()
```

```python
spb.info()
```

**а)** Сколько **различных заказов** в двух файлах вместе (дубли, попавшие в оба файла, считаются один раз)? Дальше работайте с объединённой таблицей без дублей. 

```python
# ваш код

orders = pd.concat([msk, spb]).drop_duplicates(subset=['order_id'])


orders.sample(random_state=42)


```

```python
orders.info()
```

**б)** Сколько заказов содержат **неизвестный товар** (его `product_id` отсутствует в `products.csv`)?

```python
# ваш код
(~(orders['product_id'].isin(products['product_id']))).sum()
```

**в)** Суммарная выручка по всем заказам с известными товарами, **млн руб.** Округлите до **сотых**. 

```python
full = pd.merge(orders, products, how="inner", on='product_id')
```

```python
full.info()
```

```python
full.sample(5, random_state=1)
```

```python
# ваш код
full['order_price'] = full['quantity'] * full['unit_price']
```

```python
round(full['order_price'].sum() / 1e6, 2)
```

**г)** Выручка **самой прибыльной категории** товаров, млн руб. Округлите до **сотых**. 

```python
# ваш код
round(full.groupby(by='category')['order_price'].sum() / 1e6, 2)
```

**д)** Сколько клиентов из `customers.csv` **не сделали ни одного заказа**? Учитывайте все заказы, в том числе с неизвестными товарами. 

```python
full = pd.merge(orders, products, how="outer", on='product_id')
```

```python
full.info()
```

```python
# ваш код

(~(customers['customer_id'].isin(full['customer_id']))).sum()
```

**е)** Средний «чек» (выручка одного заказа) у клиентов сегмента **VIP**, руб. Округлите до **сотых**. Берутся заказы с известными товарами.

```python
# ваш код

full = pd.merge(orders, products, how="inner", on='product_id')

full = pd.merge(full, customers, how="inner", on="customer_id")

full['order_price'] = full['quantity'] * full['unit_price']
```

```python
full.info()
```

```python
round(full[full['segment'] == 'VIP']['order_price'].mean(), 2)
```

# Задача 5. Логи мобильного приложения (datetime)

Файл `data/app_logs.csv` — действия пользователей в мобильном приложении магазина.

| столбец | описание |
|---|---|
| `user_id` | идентификатор пользователя |
| `timestamp` | дата и время действия (в виде строки) |
| `action` | тип действия |

Строки в файле **не отсортированы**.

**Сессии.** Если у пользователя между двумя соседними (по времени) действиями прошло **строго больше 30 минут**, то второе действие начинает **новую сессию**. Если перерыв ровно 30 минут или меньше — сессия продолжается. Первое действие пользователя всегда начинает его первую сессию. **Длительность сессии** — время между первым и последним действием в ней (у сессии из одного действия она равна 0).

```python
# Загрузите датасет и создайте столбец типа datetime 

app_logs = pd.read_csv('data/app_logs.csv')

app_logs['timestamp'] = pd.to_datetime(app_logs['timestamp'])
```

```python
app_logs.sample(5, random_state=43)
```

```python
app_logs.info()
```

**а)** Сколько действий было совершено **ночью** — с 00:00:00 до 05:59:59 включительно? 

```python
# ваш код

app_logs[app_logs['timestamp'].dt.hour.isin([0, 1, 2, 3, 4, 5])].shape[0]
```

**б)** В какой день недели совершено больше всего действий? Ответ — **номер дня недели**: понедельник = 0, вторник = 1, …, воскресенье = 6. 

```python
# ваш код

app_logs['dayofweek'] = app_logs['timestamp'].dt.dayofweek
app_logs.head()
```

```python
app_logs.groupby(by='dayofweek').count()
```

**в)** Найдите пользователя с **наибольшим числом сессий**. В ответе укажите его `user_id`. 

```python
# ваш код

app_logs = app_logs.sort_values(by=['user_id', 'timestamp']).reset_index(drop=True)
```

```python
app_logs.head(5)
```

```python
time_diff = app_logs.groupby('user_id')['timestamp'].diff()
time_diff

```

```python
app_logs['session_id'] = (time_diff > pd.Timedelta('30min')).cumsum()
```

```python
app_logs.groupby('user_id')['session_id'].nunique().sort_values(ascending=True)
```

```python
app_logs
```

**д)** Чему равна **средняя длительность сессии** в минутах (по всем сессиям всех пользователей)? Округлите до **сотых**. 

```python
# ваш код

sessions = app_logs.groupby('session_id')['timestamp'].agg(['min', 'max'])


sessions_durations = (sessions['max'] - sessions['min']).dt.total_seconds() / 60
sessions_durations.sort_values()


```

```python
round(sessions_durations.mean(), 2)
```
